```python
import xml.etree.ElementTree as ET

def parse_xml(file_path):
    tree = ET.parse(file_path)
    root = tree.getroot()
    
    book_data = {
        "title": root.find("title").text,
        "author": root.find("author").text,
        "chapters": []
    }
    
    # Extract chapters
    for chapter in root.find("chapters").findall("chapter"):
        chapter_title = chapter.find("title").text
        chapter_content = []
        
        # Extract medicines
        for medicine in chapter.find("content").findall("medicine"):
            med_data = {
                "name": medicine.find("name").text,
                "uses": medicine.find("uses").text,
                "dosage": {
                    "adults": medicine.find("dosage/adults").text,
                    "children": medicine.find("dosage/children").text
                },
                "side_effects": medicine.find("side_effects").text
            }
            chapter_content.append(med_data)
        
        book_data["chapters"].append({
            "title": chapter_title,
            "content": chapter_content
        })
    
    return book_data

# Parse the provided XML file
book_data = parse_xml("C:\\Users\\Osama\\AI_Lab-main\\xml_book")
book_data
```

```python
from whoosh.fields import Schema, TEXT, ID
from whoosh.index import create_in
import os

def create_search_index(book_data, index_dir="indexdir"):
    if not os.path.exists(index_dir):
        os.mkdir(index_dir)
    
    schema = Schema(title=TEXT(stored=True), content=TEXT(stored=True), medicine_name=TEXT(stored=True))
    idx = create_in(index_dir, schema)
    writer = idx.writer()
    
    for chapter in book_data["chapters"]:
        for medicine in chapter["content"]:
            writer.add_document(
                title=chapter["title"],
                content=f"{medicine['uses']} {medicine['side_effects']} {medicine['dosage']['adults']} {medicine['dosage']['children']}",
                medicine_name=medicine["name"]
            )
    
    writer.commit()

# Create the search index
create_search_index(book_data)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

def process_query(query):
    doc = nlp(query)
    # Include numerals, proper nouns, and specific keywords
    keywords = [token.text for token in doc if (token.is_alpha or token.is_digit) and not token.is_stop]
    return " ".join(keywords)

# Example usage
print(process_query("What is the content of chapter 1?"))
```
